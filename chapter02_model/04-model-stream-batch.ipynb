```python
from langchain.chat_models import init_chat_model
import os
from dotenv import load_dotenv

load_dotenv(override=True)
LLM_MODE2 = os.getenv("LLM_MODE2")
LLM_API_KEY2 = os.getenv("LLM_API_KEY2")
LLM_BASE_URL2 = os.getenv("LLM_BASE_URL2")

DEEPSEEK_MODEL = os.getenv("DEEPSEEK_MODEL")
DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")
DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")

model = init_chat_model(
    api_key=DEEPSEEK_API_KEY,
    base_url=DEEPSEEK_BASE_URL,
    model=DEEPSEEK_MODEL,
    model_provider="openai"
)

for chunk in model.stream("你好，你是谁？介绍一下"):
    print(chunk.text,end="",flush=True)
```

# batch 统一调用

```python
from rich import print as rprint
msg = [
    "你是谁？",
    "我是谁",
    "什么是人工智能"
]

res = model.batch(msg)

for item in res:
    rprint(item.text)
```

# 举例3：性能对比

使用bach

```python
# 准备多个输入
inputs = [
    "翻译成英文：春天来了",
    "翻译成英文：夏天很热",
    "翻译成英文：秋天落叶",
    "翻译成英文：冬天下雪"
]

# 批量调用
import time

start = time.time()  # 开始时间
response = model.batch(inputs) # 调用模型，批量传参
bach_time = time.time() - start  # 结束时间

print("批量调用结果：")
for i,response in enumerate(response):
    print(f"{i + 1}. {response.content}")

print(f"耗时： {bach_time:.2f}秒\n")
```

循环invoke

```python
start = time.time()  # 开始时间

# 存放返回结果
result = set()

for msg in inputs:
    res = model.invoke(msg)
    result.add(res.content)

invoke_time = time.time() - start  # 结束时间

print("批量调用结果：")
for idx,res in enumerate(result):
    print(f"{idx + 1}. {res}")

print(f"耗时： {invoke_time:.2f}秒\n")
print(f"批量调用节省{((invoke_time - bach_time) / invoke_time * 100):.1f}% ")
```
