# 使用@tool装饰器定义工具

## 1 自定义工具描述：description

举例1：

函数使用@tool装饰器装饰以后，就是一个可以被模型识别的工具了

```python
from langchain_core.tools import tool
from langchain_core.utils.function_calling import convert_to_openai_tool
from rich import print as rprint

# parse_docstring=True 开启谷歌参数文档，不开启就不显示参数信息
@tool(
    parse_docstring=True,  # 开启谷歌参数文档
    description="获取指定城市天气",  # 工具描述
    name_or_callable="getWeather"  # 重命名方法名
)
def get_weather(city:str) -> str:
    """
    返回指定城市天气

    Args:
        city:城市名
    """
    return f"{city}天气晴朗"

rprint(convert_to_openai_tool(get_weather))
```

### 3 使用pydantic模型自定义arge_schema

```python
from langchain_core.tools import tool
from langchain_core.utils.function_calling import convert_to_openai_tool
from rich import print as rprint
from pydantic import BaseModel, Field


class GetWeatherBaseModel(BaseModel):
    city: str = Field(
        default="沈阳",
        description="具体城市",
        min_length=2,
        max_length=10
    )


@tool(
    args_schema=GetWeatherBaseModel
)
def get_weather(city:str) -> str:
    """
    返回指定城市天气
    """
    return f"{city}天气晴朗"

rprint(convert_to_openai_tool(get_weather))
```
