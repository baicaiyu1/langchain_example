```python
import base64

from pathlib import Path
from langchain.chat_models import init_chat_model
from langchain_core.messages import HumanMessage

from config.config import DeepSeekModel

model = init_chat_model(
    base_url=DeepSeekModel.deepseek_base_url,
    api_key=DeepSeekModel.deepseek_api_key,
    model=DeepSeekModel.deepseek_model,
    model_provider="openai",
)

image_path = Path.cwd() / "gaosilin.jpg"

with open(image_path, "rb") as f:
    b64_str = base64.b64encode(f.read()).decode("utf-8")

message = [
    HumanMessage(
        content_blocks=[
            {"type": "text", "text": "他是谁？他的名字叫詹姆斯·高斯林"},
            {
                "type": "image",
                "base64": b64_str,        # 纯 base64，不带前缀
                "mime_type": "image/jpeg", # jpg → jpeg
            },
        ]
    )
]

print(model.invoke(message))
```
