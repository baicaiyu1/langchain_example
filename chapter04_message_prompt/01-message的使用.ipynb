```python
import os

from dotenv import load_dotenv
from langchain.chat_models import init_chat_model
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
from config.config import DeepSeekModel
# load_dotenv(override=True)
#
# DEEPSEEK_MODEL = os.getenv("DEEPSEEK_MODEL")
# DEEPSEEK_API_KEY = os.getenv("DEEPSEEK_API_KEY")
# DEEPSEEK_BASE_URL = os.getenv("DEEPSEEK_BASE_URL")

model = init_chat_model(
    model_provider="openai",
    api_key=DeepSeekModel.deepseek_api_key,
    base_url=DeepSeekModel.deepseek_base_url,
    model=DeepSeekModel.deepseek_model
)
```

```python
# json格式消息
message = [
    {'role':'system','content':'你是一个ai大模型应用开发专家'},
    {'role':'user','content':'讲一下模型应用开发需要学什么？'},
    {'role':'assistant','content':"""
    模型应用开发，重点不是“从零训练大模型”，而是**用大模型 API 或开源模型，结合数据、工具、工作流，做出可靠可上线的 AI 应用**。需要学的东西可以分成 6 层。\n\n## 1. 工程基础\n这是底座，不用很深，但必须会：\n- **Python**：异步、类型注解、Pydantic、包管理\n- **后端 API**：FastAPI / Flask，HTTP、SSE 流式输出、WebSocket\n- **数据存储**：PostgreSQL、Redis、对象存储\n- **工程工具**：Git、Linux、Docker、日志、环境变量\n- **前端基础**：至少能对接 API，理解流式聊天交互\n\n## 2. 大模型基础认知\n不要求推导 Transformer，但要理解：\n- Token、上下文窗口、Embedding、注意力机制\n- 预训练、SFT、RLHF/DPO、LoRA 微调\n- 推理参数：temperature、top_p、max_tokens\n- 模型能力边界：幻觉、知识截止、上下文遗忘、推理不稳定\n- API 调用：OpenAI、Claude、Qwen、DeepSeek、本地 Ollama/vLLM\n\n## 3. 核心应用技术\n这是模型应用开发的主战场。\n\n**提示与上下文工程**\n- 系统提示、少样本、思维链、结构化输出\n- JSON Mode、Function Calling、工具调用\n- 上下文压缩、记忆管理、多轮对话状态\n\n**RAG 知识增强**\n- 文档解析、切块、Embedding、向量数据库\n- 向量检索、关键词检索、混合检索、重排 Rerank\n- 查询改写、多路召回、引用溯源\n- 常用：Chroma、Milvus、PGVector、Elasticsearch、LlamaIndex\n\n**Agent 与工作流**\n- ReAct、Plan-and-Execute、工具调用、反思\n- 单 Agent、多 Agent、LangGraph、Dify、Coze\n- MCP、浏览器、代码执行、数据库、API 工具\n- 记忆：短期记忆、长期记忆、向量记忆\n\n## 4. 评估、可观测与安全\n这是从 Demo 到产品的关键。\n- 离线评估：准确率、召回率、忠实度、相关性\n- 在线评估：A/B 测试、用户反馈、人工评估\n- LLM-as-a-Judge、RAGAS、LangSmith、Phoenix\n- 成本、延迟、Token 消耗监控\n- 安全：Prompt 注入、越狱、数据泄露、PII、权限、审计、合规\n\n## 5. 部署与工程化\n- FastAPI 服务、流式输出、并发、限流、缓存、降级\n- Docker、K8s、云服务、CI/CD\n- 开源模型部署：vLLM、TGI、Ollama、量化\n- 微调：LoRA、QLoRA、PEFT、指令数据构造\n- 成本优化：小模型路由、缓存、批处理、蒸馏\n\n## 6. 产品与业务思维\n模型应用开发最终要解决问题：\n- 选场景：客服、知识库、数据分析、写作、审核、Copilot\n- 设计人机协同，而不是追求全自动\n- 理解 ROI、用户留存、数据飞轮\n- 把模型能力封装成稳定产品\n\n## 推荐学习顺序\n1. Python + API 调用 + Prompt\n2. 做一个聊天机器人 / 文档总结工具\n3. 学 RAG，做企业知识库问答\n4. 学 Function Calling + Agent，做工具型助手\n5. 学评估、日志、成本监控\n6. 学 FastAPI + Docker + 流式部署\n7. 最后再学微调、vLLM、多 Agent\n\n## 一句话总结\n模型应用开发 = **工程能力 + 模型理解 + 上下文/知识/工具编排 + 评估迭代 + 产品场景**。\n\n如果只记三个关键词：**上下文、工具、评估**。能把这三件事做好，就已经超过大多数只会写 Prompt 的人了。
    """},
    {'role':'user','content':'我刚才问了什么问题'},
]
result = model.invoke(message)
print(result)
```
