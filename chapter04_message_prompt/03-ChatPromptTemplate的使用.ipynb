# 2. ChatPromptTemplate的调用

### 2.1 invoke()

@param: dict
@return: ChatPromptValue

举例：

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的ai助手,你的名字是{name}"),
    ("human","你好，我想问一下1+1=？"),
    ("ai","1+1=3"),
    ("human","{user_input}"),
])

result = chat_prompt_template.invoke({"name":"小智","user_input":"我刚才问什么了？"})
print(type(result))
```

### 2.2 format()

@param: 普通变量
@return: str

举例:


```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的ai助手,你的名字是{name}"),
    ("human","你好，我想问一下1+1=？"),
    ("ai","1+1=3"),
    ("human","{user_input}"),
])

result = chat_prompt_template.format(name="小智",user_input="我刚才问的什么？")
print(type(result))
```

### 2.3 format_messages()

@param: 普通变量
@return: list

举例

```python
from langchain_core.prompts import ChatPromptTemplate

chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的ai助手,你的名字是{name}"),
    ("human","你好，我想问一下1+1=？"),
    ("ai","1+1=3"),
    ("human","{user_input}"),
])

result = chat_prompt_template.format_messages(name="小智",user_input="我刚才问的什么？")
print(type(result))
```

### 2.4 结合大模型使用提示词模板

```python
from config.config import DeepSeekModel
from langchain.chat_models import init_chat_model
from rich import print as rprint

# ========= 初始化大模型 ========
model = init_chat_model(
    base_url=DeepSeekModel.deepseek_base_url,
    api_key=DeepSeekModel.deepseek_api_key,
    model=DeepSeekModel.deepseek_model,
    model_provider="openai"
)

# ========= 初始化提示词模板 ========
chat_prompt_template = ChatPromptTemplate.from_messages([
    ("system","你是一个友好的ai助手,你的名字是{name}"),
    ("human","你好，我想问一下1+1=？"),
    ("ai","1+1=3"),
    ("human","{user_input}")
])

#  ========== 组装提示词模板 =========
prompt_result = chat_prompt_template.invoke({"name":"小智","user_input":"我刚才问什么了？"})

# 调用大模型 传入提示词模板
model_response = model.invoke(prompt_result)

rprint(model_response)
```

# 4 更丰富的初始化参数类型

### 4.1 字符串列表


```python
from config.config import DeepSeekModel
from langchain.chat_models import init_chat_model
from rich import print as rprint

# ========= 初始化大模型 ========
model = init_chat_model(
    base_url=DeepSeekModel.deepseek_base_url,
    api_key=DeepSeekModel.deepseek_api_key,
    model=DeepSeekModel.deepseek_model,
    model_provider="openai"
)

chat_prompt_template = ChatPromptTemplate.from_messages([
    "你好，请用一句话回答我" # HumanMessage(content='你好，请用一句话回答我', additional_kwargs={}, response_metadata={})
])


prompt_result = chat_prompt_template.invoke({})

rprint(prompt_result)

# rprint(model.invoke(prompt_result))
```

### 4.2 元组列表

```python
chat_prompt_template = ChatPromptTemplate([
    ('system','你是一个友好的人工智能助手'),
    ('human','你好{name}')
])

prompt_result = chat_prompt_template.invoke({'name':'小明'})

rprint(prompt_result)
```

### 4.3 字典列表

```python
chat_prompt_template = ChatPromptTemplate([
    {'role':'system','content':'你是一个助手'},
    {
        'role':'human',
        'content':'你是{name}'
    }
])

prompt_result = chat_prompt_template.invoke({'name':'助手'})
rprint(prompt_result)
```
